```python
def check_loan(age, salary, credit_score):
  if age < 21:
    print("Age is below 21")
    print("Loan rejected!!!")
  elif salary < 50000:
    print("Salary less than Rs.50,000")
    print("Loan rejected!!!")
  elif credit_score < 650:
    print("Credit Score is less than 650")
    print("Loan rejected!!!")
  else:
    print("Loan Approved")

age = int(input("What's your age? "))
salary = int(input("What's your salary? "))
credit_score = int(input("What's your Credit Score? "))

check_loan(age, salary, credit_score)
```
